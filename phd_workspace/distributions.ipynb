Beta distribution

```python


def calc_beta_distribution(x, a, b):
    """
    Calcualte beta distribution for a given a, b and x
    """
    return (x**(a-1) * (1-x)**(b-1)) / (np.math.gamma(a) * np.math.gamma(b) / np.math.gamma(a+b))

```
