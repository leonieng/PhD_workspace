```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta
```

Beta distribution

```python
def calc_beta_distribution(x, a, b):
    """
    Calcualte beta distribution for a given a, b and x
    """
    return beta.pdf(x, a, b)
# set a, b
a=0.1
b=0.1

# plot
fig, ax = plt.subplots(1, 1)

x = np.linspace(0.0,
                1.0, 100)
ax.plot(x, calc_beta_distribution(x, a, b),
       'r-', lw=1, alpha=0.6, label=f"a={a}, b={b}")
ax.set_xlabel("x")
ax.set_ylabel(r"$\mathrm{Beta}(x|a,b)$")
ax.legend(loc="upper right")
plt.show()
```
